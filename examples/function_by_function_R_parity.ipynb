## Key differences

1. **Argument order**: Python uses keyword args
2. **Sparse matrix**: scipy.sparse.csc_matrix vs R Matrix
3. **Clustering**: sklearn + leidenalg vs N2R + igraph
4. **Gene index**: Python uses int column indices, R uses strings
